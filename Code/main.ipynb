### TEST #9: Testing the Linked Deque:

```python
# Create a new empty deque.
test_deque = LinkedDeque()

# Check that the deque starts empty.
print("STARTING EMPTY:")
print("Empty:", test_deque.is_empty())
print("Size:", test_deque.size)

# Try removing from empty deque.
print("\nREMOVE FROM EMPTY:")
print("Remove front:", test_deque.remove_front())
print("Remove back:", test_deque.remove_back())
print("Size:", test_deque.size)

# Add one item to the front.
test_deque.add_to_front(10)

# Check the front, back, and size.
print("\nAFTER ADDING ONE ITEM:")
print("Front:", test_deque.get_front())
print("Back:", test_deque.get_back())
print("Size:", test_deque.size)

# Remove the only item.
test_deque.remove_back()

# Confirm the deque is empty
print("\nAFTER REMOVING THE ONLY ITEM:")
print("Empty:", test_deque.is_empty())
print("Size:", test_deque.size)

# Add items to both ends.
test_deque.add_to_back(20)
test_deque.add_to_front(10)
test_deque.add_to_back(30)

# Check the order and size.
print("\nAFTER ADDING TO BOTH ENDS:")
test_deque.display()
print("Size:", test_deque.size)

# Remove one item from both ends.
print("\nREMOVING FROM BOTH ENDS:")
print("Removed front:", test_deque.remove_front())
print("Removed back:", test_deque.remove_back())
print("Remaining front:", test_deque.get_front())
print("Size:", test_deque.size)

# Clear the deque.
test_deque.clear()

# Confirm the deque is empty
print("\nAFTER CLEARING:")
print("Empty:", test_deque.is_empty())
print("Size:", test_deque.size)

# Confirm we can reuse the deque after it was cleared.
test_deque.add_to_back(99)

print("\nAFTER REUSING:")
print("Front:", test_deque.get_front())
print("Back:", test_deque.get_back())
print("Size:", test_deque.size)
```
